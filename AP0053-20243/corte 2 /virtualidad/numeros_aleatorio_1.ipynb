```python
import csv
import random
import time

nombre_archivo = "numero_aleatorio.csv"
with open(nombre_archivo, mode='w', newline='') as archivo_csv:
    escritor = csv.writer(archivo_csv)
    escritor.writerow(["Tiempo (s)", "Número Aleatorio"])

    tiempo_inicial = time.time()

    while True:
        tiempo_actual = time.time() - tiempo_inicial
        if tiempo_actual > 10:
            break

        numero_aleatorio = random.uniform(0, 1)
        escritor.writerow([f"{tiempo_actual:.2f}", numero_aleatorio])
        time.sleep(0.1)

print(f"Los datos se han guardado en el archivo {nombre_archivo}.")
```
