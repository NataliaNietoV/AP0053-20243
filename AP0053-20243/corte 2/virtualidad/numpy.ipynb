```python
#MATRIZ PRO MEDIO DE LISTAS

list_of_lists = [[1,2,3],[4,5,6],[7,8,9]]
matrix = np.array(list_of_lists)
print (matrix)
```

```python
print (matrix[0][0]) # 1,2,3 es 0 e igual con las filas
print (matrix[0])
print (matrix[1])
print (matrix[2])

```

```python
print (np.size(matrix)) #3*3
```

```python
print ('matrix')
print (matrix)
matrix [0][1] = 9
print ("updated matrix")
print (matrix)
```

```python
##### CODIGO DE ACTIVIDAD ####
list_of_lists = [[0,0,0],[0,0,0],[0,0,0]]
matrix4 = np.array(list_of_lists)
print (matrix4)
```

```python
print ('matrix4')
print (matrix4)
matrix4 [1,:] = 4
#matrix4 [1][0] = 4
#matrix4 [1][1] = 4
#matrix4 [1][2] = 4
print ("updated matrix")
print (matrix4)

```
